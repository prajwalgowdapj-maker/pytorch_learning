Reshaping,stacking,squeezing and ubsqueezing

```python
x=torch.arange(0,100,10)
x,x.shape
```

```python
x.reshape(2,5)
```

```python
#if i use view, both z and x share same memory
z=x.view(2,5)
z
```

```python
#stack tensor on top of each other
x_stacked=torch.stack([x,x,x],dim=0)
x_stacked
```

```python
x=torch.rand([1,1,10])
x,x.shape
```

```python
#torch.squeeze()-removes all single dimensions from a target tensor
s=x.squeeze()
s
```

```python
s.shape
```

```python
# torch.unsqueeze()-adds a single dimension to a target tensor at a specific dim
u=s.unsqueeze(dim=1)
u,u.shape
```

```python
#torch.permute-rearrange the dimensions of a target tensor in a specified order
image=torch.rand(224,224,3)
image.shape
```

```python
p=image.permute(1,0,2)
p.shape
```

Reproducible tensors

```python
#Set the random seed
RANDOM_SEED=42
torch.manual_seed(RANDOM_SEED)
random_tensor_A=torch.rand(4,5)

torch.manual_seed(RANDOM_SEED)
random_tensor_B=torch.rand(4,5)

random_tensor_A,random_tensor_B
```
